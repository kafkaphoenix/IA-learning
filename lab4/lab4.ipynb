# Deep learning

Deep learning is a group of machine learning techniques that teach computers to learn by example. It is a subset of artificial intelligence (AI) and machine learning (ML) that uses neural networks with many layers (hence "deep") to model and understand complex patterns in data.

## Data Representation

In deep learning, data is typically represented in the form of tensors, which are multi-dimensional arrays. The way data is represented can significantly impact the performance of deep learning models. Common data representations include:
- **Images**: Represented as 3D tensors (height, width, channels).
- **Audio**: Represented as 1D or 2D tensors (waveforms or spectrograms).
- **Tabular Data**: Represented as 2D tensors (rows and columns).

> We don't have a phase of feature extraction as in traditional ML. Instead, deep learning models learn to extract features automatically from raw data during training.

However, due to the data now being more complex, the models also become more complex, requiring more computational resources and larger datasets to train effectively.

- Data relevance: Data has to be relevant to the task at hand.
- Labels: Labeled data has to be accurate and consistent.
- Format: All data has to be in the same format (e.g., image size, color channels).
- Quantity: Deep learning models typically require large amounts of data to perform well and balanced datasets (Data augmentation can help).

> For example for object detection in images, we need images with bounding boxes and labels for each object in the image. But for instance segmentation, we need categorical labels and coordinates for each object we want to detect.

> Or for example when classifying images of animals, we need images of the specific animals we want to classify, and not images of other objects or animals. And the format has to be the same, for example all images have to be in color or all in grayscale. Or all images have to be of the same size.

## Data Augmentation

Data augmentation techniques can be used to artificially increase the size of the dataset by applying transformations such as rotation, scaling, cropping, black and white conversion, and flipping to existing data samples.

## Neural Networks

Neural networks are the backbone of deep learning. They consist of layers of interconnected nodes (neurons) that process and learn from data. Common types of neural networks include:
- **Feedforward Neural Networks (FNNs)**: The simplest type of neural network where information flows in one direction from input to output.
- **Convolutional Neural Networks (CNNs)**: Primarily used for image data, CNNs use convolutional layers to automatically and adaptively learn spatial hierarchies of features.
- **Recurrent Neural Networks (RNNs)**: Designed for sequential data, RNNs have connections that form directed cycles, allowing them to maintain a 'memory' of previous inputs.

Neurons operate as individual computational units, but their behavior is controlled by the activation function and the bias. The activation function determines when and how strongly a neuron activates, introducing non-linearity into the model. The bias is a constant added to the weighted sum of inputs, allowing the neuron to shift its activation threshold and learn more flexible patterns.

A neuron receives inputs that are multiplied by weights, sums them, adds the bias, and then applies an activation function such as sigmoid, tanh, or ReLU to produce an output. This process enables neural networks to model complex, non-linear relationships in data.

### Activation Functions
- **Sigmoid**: Outputs values between 0 and 1, useful for binary classification.
- **Tanh**: Outputs values between -1 and 1, often used in hidden layers.
- **ReLU (Rectified Linear Unit)**: Outputs the input directly if positive; otherwise, it outputs zero. It is widely used due to its simplicity and effectiveness in deep networks. Used in hidden layers.
- **Softmax**: Used in the output layer for multi-class classification, it converts logits into probabilities.

### Layers
- **Input Layer**: The first layer that receives the input data. It has neurons corresponding to the features of the input data.
- **Hidden Layers**: Intermediate layers that process inputs from the previous layer and pass outputs to the next layer.
- **Output Layer**: The final layer that produces the output of the network.

> We don't know exactly how many hidden layers or neurons per layer we need. This is something that we have to experiment with and tune based on the specific problem and dataset. Also we don't know what happens inside the hidden layers, they are like a black box that we can't interpret easily. For example, for classifying two categories, we would need two output neurons, one for each category. But for regression problems, we would need only one output neuron that gives us a continuous value.

### Training Neural Networks

Training neural networks involves adjusting the weights and biases of the neurons to minimize the difference between the predicted outputs and the actual targets. This is typically done using a process called backpropagation combined with an optimization algorithm like stochastic gradient descent (SGD) or Adam.

The training process includes:
- **Forward Pass**: Input data is passed through the network to obtain predictions (with the current weights and biases). We get the error by comparing the predictions to the actual targets using a loss function (e.g., mean squared error for regression, cross-entropy for classification).
- **Backward Pass**: We use the gradient of the loss function with respect to each weight and bias to update them in a way that reduces the loss. This is done using the chain rule of calculus to propagate the error backward through the network.

### Basic Architectures - FNN, RNN, CNN

We have different architectures for different types of data and tasks:
- **Feed forward Neural Networks (FNNs)**: Used for image classification, regression, fraud detection, simple networks, NLP tasks. (it could be not the most efficient for one task but it works for most of them)
- **Recurrent Neural Networks (RNNs)**: Used for sequential data like time series, speech recognition, language modeling. (Networks with memory)
- **Convolutional Neural Networks (CNNs)**: Used for image data, object detection, image segmentation, video analysis. (Networks with filters)


> Data alters the architecture choice. For example, CNNs are preferred for image data due to their ability to capture spatial hierarchies, while RNNs are better suited for sequential data like text or time series.

#### RNN Variants
With RNNs we don't talk about layers but about cells. There are different types of RNN cells that are designed to handle specific challenges in sequential data processing:
- **LSTM (Long Short-Term Memory)**: A type of RNN that can learn long-term dependencies and is effective for tasks like language modeling and time series prediction. It uses gates to control the flow of information and maintain memory over long sequences. It has lot of parameters so it is computationally expensive.
- **GRU (Gated Recurrent Unit)**: A simplified version of LSTM that is computationally more efficient while still capturing long-term dependencies. It works well for similar tasks as LSTM but with fewer parameters.
> Choosing between LSTM and GRU often depends on the specific task and dataset, as well as computational resources.
- **ANNs (Attention Neural Networks)**: These networks use attention mechanisms to focus on specific parts of the input sequence, improving performance in tasks like machine translation and text summarization. It does not have memory like RNNs but it can capture relationships between distant elements in the input data. (it needs past data to work, for example stock values, electric bills, etc).

#### CNN

- RRNN Number of input neurons: Corresponds to the number of features in the input data (e.g., pixels in an image). An image of size 1920x1080 with 3 color channels (RGB) would require 1920 * 1080 * 3 = 6,220,800 input neurons. Really computationally expensive. So it is better to use CNNs for image data.
- Convolutional Layers: Apply filters to the input data to extract features. Each filter slides over the input data, performing a convolution operation that produces feature maps.
- Pooling Layers: Reduce the spatial dimensions of the feature maps, retaining important information while reducing computational load. Common pooling methods include max pooling and average pooling.
- Fully Connected Layers: After several convolutional and pooling layers, the high-level features are flattened and passed through fully connected layers to make final predictions.

### Model Evaluation

The evaluation of deep learning models is crucial to ensure they perform well on unseen data. Common evaluation metrics include accuracy(it needs balanced data), precision, recall, F1-score, confusion matrix (it summarizes the performance of a classification model) for classification tasks, and mean squared error (MSE), mean absolute error (MAE) and R-squared for regression tasks. Cross-validation techniques can also be employed to assess model performance more robustly.

> Similar to machine learning, but with larger datasets and more complex models. We split the data into training, validation, and test sets to evaluate the model's performance and generalization ability.

### Use Cases and Limitations

Deep learning has been successfully applied in various domains, including:
- Computer Vision: Image classification, object detection, image segmentation.
- Natural Language Processing: Machine translation, sentiment analysis, text generation.
- Speech Recognition: Voice assistants, transcription services.

However, deep learning also has limitations:
- Data Requirements: Requires large amounts of labeled data for effective training.
- Computational Resources: Training deep learning models can be resource-intensive, requiring powerful GPUs or TPUs.
- Interpretability: Deep learning models are often considered "black boxes," making it difficult to understand how decisions are made.
- Overfitting: Deep learning models can overfit to training data, especially when the dataset is small or not diverse enough. We have mechanmisms to reduce it like k-fold cross validation, dropout, regularization, early stopping, etc.
- Degradation: Performance may degrade when applied to data that differs significantly from the training data. It requires continuous monitoring and updating to maintain effectiveness.
- Missing comprehension: Deep learning models may struggle to understand context or nuances in data, leading to errors in complex tasks.
