# Final project

Given the theorical nature of this lab, the final project consists of answering the following questios related to deep learning concepts:

- Question 1: Is correct to apply the same evaluation metrics for a model detecting cancers in medical images than for a model classifying music genres? Justify your answer.

- Question 2: Deep learning models need large amounts of data to perform well. However, lot of times we do not have access to such public datasets. Would be ethical to scrape data from the web to create our own dataset? Alternatives?

- Question 3: Most libraries for deep learning (TensorFlow, PyTorch...) use GPUs to speed up the training process. Why are GPUs more efficient than CPUs for this task?

Answers:
- Answer 1: No, it is not correct to apply the same evaluation metrics for a model detecting cancers in medical images and for a model classifying music genres. In medical image analysis, the cost of false negatives (failing to detect cancer) is much higher than false positives (incorrectly identifying healthy tissue as cancerous). Therefore, metrics like sensitivity (recall) and specificity are more critical. In contrast, for music genre classification, accuracy might be a more appropriate metric since the consequences of misclassification are less severe.
- Answer 2: Scraping data from the web to create a dataset can raise ethical and legal concerns, especially if the data is copyrighted or contains personal information. It is essential to respect intellectual property rights and privacy laws. Alternatives include using publicly available datasets, collaborating with institutions that can provide data, or generating synthetic data using techniques like data augmentation or generative adversarial networks (GANs).
- Answer 3: GPUs (Graphics Processing Units) are more efficient than CPUs (Central Processing Units) for deep learning tasks because they are designed to handle parallel processing. Deep learning involves performing a large number of matrix multiplications and other operations that can be parallelized. GPUs have thousands of smaller cores that can perform many calculations simultaneously, making them well-suited for the high computational demands of training deep learning models. In contrast, CPUs have fewer cores optimized for sequential processing, which makes them less efficient for these tasks.
