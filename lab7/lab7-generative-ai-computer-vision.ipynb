# Generative IA in Computer Vision

## Generative AI in Computer Vision
Generative AI has made significant strides in the field of computer vision, enabling the creation of realistic images, videos, and 3D models. Techniques such as Generative Adversarial Networks (GANs) and Variational Autoencoders (VAEs) have been widely adopted for various applications, including image synthesis, style transfer, and data augmentation. Recent advancements in diffusion models have further enhanced the quality and diversity of generated visual content, making them a popular choice for tasks such as image generation and super-resolution.

## Diffusion Models
Diffusion models are a class of generative models that learn to generate data by reversing a gradual noising process. The core idea behind diffusion models is to start with random noise and iteratively refine it to produce a coherent output.

### Key Concepts
- **Latent Space**: Diffusion models often operate in a latent space, where the data is represented in a lower-dimensional form. This allows for more efficient learning and generation.
- **Number of Steps**: The diffusion process involves a series of steps, where noise is gradually added to the data during training and removed during generation. The number of steps can significantly impact the quality of the generated output. Final steps can add fine details to the generated images or even change the overall structure.

### Training Process
1. **Forward Process**: During training, the model learns to add noise to the data in a controlled manner over several steps.
2. **Reverse Process**: The model is trained to reverse the noising process, effectively learning to denoise the data step by step.



## Stable Diffusion
Stable Diffusion is a popular diffusion model architecture that has gained attention for its ability to generate high-quality images from text prompts. It combines the principles of diffusion models with techniques such as attention mechanisms and transformer architectures to enhance the generation process. It's open-source.

### Hugging Face SD Architecture 1.4
https://huggingface.co/blog/stable_diffusion

![stable diffusion architecture](images/stable_diffusion_architecture.png)


### Other SD Architectures
- **Stable Diffusion XL (SDXL)**: It adds a new step of refinement to the diffusion process, resulting in even higher quality images with more intricate details and VAE-based decoder for increasing the resolution of generated images.
- **Stable Diffusion XL Turbo**: A lighter and faster version of Stable Diffusion, optimized for speed and efficiency while maintaining good image quality. The refiner step also increase the resolution of generated images, not needing a VAE-based decoder.
- **Stable Cascade**: It uses multiple diffusion models in a cascade to progressively refine the generated images, leading to higher quality outputs. Each model in the cascade focuses on different aspects of the image generation process, such as coarse structure and fine details.

### Task a SD Can Solve
- Text-to-Image Generation: Creating images based on textual descriptions.
- Image Inpainting/Outpainting: Filling in missing parts of an image or extending the image beyond its original borders.
- Image Super-Resolution: Enhancing the resolution of images (Upscaling).
- Image-to-Image Translation: Transforming images from one style or domain to another (e.g., turning sketches into realistic images).

### ControlNet
ControlNet is an extension of Stable Diffusion that allows for more precise control over the generated images by incorporating additional input conditions, such as edge maps, depth maps, or segmentation masks. This enables users to guide the image generation process more effectively, resulting in outputs that better align with their desired specifications.

## Training a Stable Diffusion Model

There are several popular methods for fine-tuning Stable Diffusion models to learn new concepts, each with its own strengths and trade-offs. Here are four widely used techniques:

### **DreamBooth**
Paper: https://arxiv.org/abs/2208.12242
DreamBooth is a full fine-tuning approach that teaches a Stable Diffusion model a **new specific concept** (e.g., a person, pet, object, or style) using a small set of images and a unique identifier token.

**Pros**

* Excellent at learning **specific identities** (faces, characters, objects)
* High fidelity and strong concept retention
* Works well with very few images (3–20)
* Produces consistent results across prompts

**Cons**

* Computationally expensive (VRAM, time, storage)
* Risk of **overfitting** or style leakage if not regularized
* Less modular: the whole model must be loaded
* Harder to combine multiple DreamBooth concepts

**Best for:** Photorealistic subjects, characters, or cases where identity accuracy matters most.

---

### **Textual Inversion**
Paper: https://arxiv.org/abs/2208.01618

Textual Inversion learns a **new token embedding** that represents a concept in the model’s latent space, without modifying the model weights.

**Pros**

* Very lightweight (small files, fast training)
* No risk of damaging the base model
* Easy to share and reuse
* Can be combined freely with other methods

**Cons**

* Limited expressiveness for complex concepts
* Struggles with faces, fine details, or structural consistency
* Highly prompt-dependent
* Less control over pose, lighting, or composition

**Best for:** Styles, textures, aesthetics, or abstract concepts rather than specific subjects.

---

### **LoRA (Low-Rank Adaptation)**
Paper: https://arxiv.org/abs/2106.09685

LoRA fine-tunes the model by injecting **low-rank weight updates** into specific layers, allowing efficient adaptation without retraining the full network.

**Pros**

* Excellent balance between quality and efficiency
* Small file size and fast training
* Easy to mix multiple LoRAs at inference time
* Minimal risk to the base model
* Works well for both **style and character learning**

**Cons**

* Slightly less expressive than full DreamBooth
* Requires careful layer and rank selection
* Can conflict with other LoRAs if poorly trained

**Best for:** Most use cases today—characters, styles, poses, outfits, and modular fine-tuning.

---

### **HyperNetworks**
Paper: https://arxiv.org/abs/1806.05393

HyperNetworks train a secondary network that dynamically generates weights for parts of the main model based on input conditions.

**Pros**

* Allows **dynamic adaptation** during inference
* Can capture stylistic transformations cleanly
* Modular and reusable across models

**Cons**

* Generally lower quality than LoRA or DreamBooth
* More complex training and tuning
* Largely superseded by LoRA in practice
* Limited community support and tooling

**Best for:** Experimental setups or research-driven workflows rather than production use.
