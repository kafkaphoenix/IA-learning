# 5. Strategies of prompting and fine-tuning and last trends of language models

## Prompting Strategies

1. Prompt engineering involves designing prompts to elicit desired responses from language models. Common strategies include:
    - **Zero-shot prompting**: Asking the model to perform a task without any examples.
    - **Few-shot prompting**: Providing a few examples in the prompt to guide the model.
    - **Chain-of-thought prompting**: Encouraging the model to reason through a problem step-by-step.
2. RAG (Retrieval-Augmented Generation) combines retrieval of relevant documents with generation capabilities of language models to improve accuracy and relevance.
3. Fine-tuning involves training a pre-trained language model on a specific dataset to adapt it to a particular task or domain. For example image generation with the dataset of an artist's works.

> You can use the three strategies together.

## Last Trends in Language Models

- Multi-modal models: These models can process and generate content across multiple modalities, such as text, json, images, and audio (they don't convert audio to text, instead they work directly with audio data).
- Access to external knowledge: Some models can access external databases, APIs, or the internet to retrieve up-to-date information during generation or do complex tasks (Now models have structured data understanding capabilities).
- Agentic workflows: Models can act as agents that autonomously perform tasks by interacting with tools, APIs, or other models to achieve complex goals. For example, using [langgraph](https://github.com/langchain-ai/langgraph) to create workflows with multiple models and tools.
- Inference-scaling or test-time compute: Until now the models were improved increasing the number of parameters (size of the model) or the dataset used for training. An alternative approach is to scale computation at inference time, giving models more time or steps to reason. Techniques include self-refinement, chain-of-thought reasoning (requesting it or giving an example of how to do it), and explicit planning before answering, which can significantly improve performance without increasing model size.

> An example of this is processing emails where the model extracts relevant information and generates a response based on the extracted data in a structured format. For this we would give the model examples of email and desired outputs and also we would create a plan for the model to follow before processing the email. We could also give the model some methods to improve its performance at inference time like one to transform dates to months.

> Or when processing movie reviews, we could clean the descriptions first, then extract relevant information like actors, director, genre, and give to the models some examples of how to do it, and finally generate a summary or analysis based on the extracted data.
