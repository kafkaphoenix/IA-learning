## Use cases

Right now AI is being used in many industries and sectors but is not common knowledge, so lot of companies they need to be educated about the possibilities of AI. For that reason, some type the MLE has to help to identify use cases where AI can be applied to solve business problems. The steps to identify use cases are:
1. Define use cases: Understand the business objectives and challenges. Analyze existing processes and workflows.
2. Business value analysis: Identify areas where AI can add value.
3. Acceptance criteria: Define success metrics and KPIs.
4. Viability and cost-benefit analysis: Prioritize use cases based on feasibility and impact.

> Best AI is not AI. The best AI is no AI at all. Use AI only when it adds value to the business and solves real problems.

## No model training

Currently, most of the AI applications in production do not involve model training. The models are pre-trained and then deployed to production. This is because training models require a lot of resources, time and expertise. Instead, most of the work is done in inference, which is the process of using a trained model to make predictions on new data.

For example using a pre-trained model from Hugging Face or TensorFlow Hub and then deploying it to production. This way, we can leverage the power of AI without the need to train models from scratch.

But we still the phase of cleaning and preparing the data, model selection, hyperparameter tuning, prompt engineering, optional rag, chain-of-thoughts and guards to avoid hallucinations, evaluation and validation. Depending of the project we can done fine tuning to adapt the model to our specific use case, like using LoRa with a personalized dataset before requesting the model to generate images.

![adapting-llm](images/adapting-llm.png)

> Thus, the focus shifts from training models to effectively utilizing and adapting pre-trained models to meet specific business needs.

## Embeddings

Embeddings are a way to represent data in a high-dimensional space. They are used to capture the semantic meaning of data, such as text, images, or audio. In AI applications, embeddings are often used for tasks such as:
- Similarity search
- Clustering
- Classification
- Recommendation systems
- Topic modeling: Identifying topics in large text corpora.
- Retrieval-augmented generation (RAG)
